```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler
```

```python
file_path='/content/preprocessing_practice_dataset.csv'
preprocessing_data =pd.read_csv(file_path)
```

```python
numerical_columns=['Age','Experience_Years','Monthly_Salary']
```

```python
preprocessing_data.dropna(subset=numerical_columns, inplace=True)
```

```python
minmax_scaler=MinMaxScaler()
standard_scaler=StandardScaler()
```

```python
preprocessing_data['Age_MinMax']=minmax_scaler.fit_transform(preprocessing_data[['Age']])
preprocessing_data['Experience_Years_MinMax']=minmax_scaler.fit_transform(preprocessing_data[['Experience_Years']])

preprocessing_data['Age_Standard']=standard_scaler.fit_transform(preprocessing_data[['Age']])
preprocessing_data['Experience_Years_Standard']=standard_scaler.fit_transform(preprocessing_data[['Experience_Years']])
```
